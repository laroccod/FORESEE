## 1. Specifying the Model

The phenomenology of the dark states with MDM can be described by the following Lagrangian

\begin{equation}
 \mathcal{L} = - \textcolor{red}{m_\chi} \bar \chi \chi + \dfrac{1}{2} \textcolor{red}{\mu_\chi} \bar{\chi} \sigma^{\mu\nu} \chi F_{\mu \nu}
\end{equation}

with the dark state mass $m_\chi$ and the MDM $\mu_\chi$ as free parameters. For the search for dark states with MDM at forward experiments we need to know i) the *production rate*, and ii) the *interaction rate*. All these properties are specified in the `Model` class. We initialize it with the name of the model as argument. 

```python
energy = "14" # CM energy of LHC in unit of TeV
modelname="MDM"
model = Model(modelname)
```

```python
# Builder parameters: defaults exposed by build.py / load_model("EM-MDM").
nsample_2body = 2000
nsample_3body = 2000
generators_light = ['EPOSLHC', 'SIBYLL', 'QGSJET'][:1]
# Quarkonia (J/psi, psi(2S), Upsilon(nS)) use Pythia8-Monash directly.
```

**Production** In the simplest case, the dark state with MDM is produced via two body decays of vector mesons $M \to \bar \chi \chi$ with $M = \rho, \omega, \phi, J/\psi, \psi(2S)$ and $\Upsilon(nS)$. The corresponding branching fractions are given by

\begin{equation}
\frac{\text{BR}(M \!\to\! \chi\bar\chi)}{\text{BR}(M \!\to\! ee) } 
= \dfrac{\mu_\chi^2 m_M^2}{8\pi\alpha} 
\frac{(m_M^2 \!+\!8m_\chi^2)(m_M^2 \!-\! 4m_\chi^2)^{1/2}} {(m_M^2 \!+\! 2m_e^2)(m_M^2 \!-\! 4m_e^2)^{1/2}} \ . 
\end{equation}

The benchmark value of $\mu_\chi$ is taken to be Bohr magneton $\mu_B = 296.342\,{\rm GeV}^{-1}$.

**Interaction Rate:** 
Dark states with EDMs can scatter off of electrons resulting in electron recoil signatures. Following [1811.04095](https://arxiv.org/abs/1811.04095), the differential scattering rate with respect to the recoil energy is given by 
\begin{equation}
 \frac{d\sigma}{dE_R} = \left(\frac{d_\chi}{\mu_B}\right)^2 \frac{\alpha \mu_B^2}{E_{\chi }^2  } \times \left(
                          \frac{ m_\chi^2}{2 m_e} +  \frac{E_\chi^2}{E_R }  \right)
\end{equation}

```python
model.set_dsigma_drecoil_1d(
    dsigma_der="296.342**2/137./energy**2 * (energy**2/recoil + mass**2/(2*0.000511))", 
    recoil_max = "2 * 0.000511 * (energy**2-mass**2) / (0.000511*(2*energy+mass) + mass**2)",
    coupling_ref=1
)
```

For our benchmark point, let us now look at how many particles interact inside the detector volume.

```python
setupnames  =  generators_light

_, _, momenta, weights = foresee.get_events_interaction(
    mass = mass, 
    energy=energy, 
    couplings= [coupling], 
    nsample = 1
)

for isetup, setup in enumerate(setupnames):
    print("Expected number of events for "+setup+":", round(sum(weights[0][:,isetup]),7))
```

Let us plot the resulting energy distribution.

```python
fig = plt.figure(figsize=(7,5))
ax = plt.subplot(1,1,1)
energies = [p.e for p in momenta], 
for isetup, setup in enumerate(setupnames):
    ax.hist(energies, weights=weights[0][:,isetup], bins=np.logspace(2,4, 20+1), histtype='step', label=setup) 
ax.set_xscale("log")
ax.set_xlim(1e2,1e4) 
ax.set_xlabel("E [GeV]") 
ax.set_ylabel("Number of Events per Bin") 
ax.legend(frameon=False, labelspacing=0, fontsize=14, loc='upper left')
os.makedirs(f"figures/{modelname}", exist_ok=True)
plt.savefig(f"figures/{modelname}/E_distribution_{modelname}.pdf", bbox_inches="tight")
plt.show()
```

Let us now scan over various masses and couplings, and record the resulting number of events. Note that here we again consider the FLArE configuration, which we set up before.

```python
configuration = 'FLArE'

condition = f"np.sqrt(p**2 + mass**2) > 100"
if not all(os.path.exists(f"model/results/{energy}TeV_{configuration}_{label}.npy") for label in setupnames):
    
    list_nevents = {label:[] for label in setupnames}
    for mass in masses:
        _, nevents, _, _ = foresee.get_events_interaction(mass = mass, energy=energy, couplings= couplings, nsample = 5, preselectioncuts = condition)
        for i,label in enumerate(setupnames): list_nevents[label].append(nevents.T[i])  
    
    #save results
    os.makedirs(f"model/results", exist_ok=True)
    for label in setupnames: 
        result = np.array([masses,couplings,list_nevents[label]], dtype='object')
        np.save("model/results/"+energy+"TeV_"+configuration+"_"+label+".npy",result)
```
